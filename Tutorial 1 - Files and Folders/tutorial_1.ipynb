## 1. the `with ... open as` pattern

```python
with open('code.py', 'w') as f:
    data = "print('hello')"
    f.write(data)
```

## 2. working with directories

### viewing directories

```python
import os
```

```python
print(os.listdir(), "\n")
```

```python
with os.scandir() as entries:
    for entry in entries:
        print(entry.name)
```

```python
from datetime import datetime
from os import scandir
```

```python
def convert_date(timestamp):
    d = datetime.fromtimestamp(timestamp)
    formated_date = d.strftime('%d %b %Y')
    return formated_date

def get_files():
    dir_entries = scandir('../Tutorial 1 - Files and Folders/')
    for entry in dir_entries:
        if entry.is_file():
            info = entry.stat()
            print(f'{entry.name}\t Last Modified: {convert_date(info.st_mtime)}')
```

```python
get_files()
```

## 3. traversing directories and processing files

```python
for dirpath, dirnames, files in os.walk('.'):
    print(f'Found directory: {dirpath}')
    for file_name in files:
        print(file_name)
```

## 4. temporary files

```python
from tempfile import TemporaryFile
```

```python
fp = TemporaryFile('w+t')
fp.write('Hello universe!')
```

```python
fp.seek(0)
data = fp.read()

print(data)
```

```python
fp.close()
```

```python
with TemporaryFile('w+t') as fp:
    fp.write('Hello universe!')
    fp.seek(0)
    data = fp.read()
    print(data)
    fp.read()
```

## 6. copying, moving and renaming files and directories

```python
with open('file1.txt', 'w') as f:
    data = 'hello world'
    f.write(data)

with open('file2.txt', 'w') as f:
    data = 'hello world'
    f.write(data)

os.mkdir("folder1")
os.mkdir("folder2")
```

### copying

```python
src = 'file1.txt'
dst = 'folder1/'
shutil.copy(src, dst)
```

```python
shutil.copytree("folder1", "folder1_backup")
```

### moving

```python
shutil.move('folder1_backup/', 'backup/')
```

### renaming

```python
os.rename('file2.txt', 'file3.txt')
```

```python
os.rename("folder2", "folder3")
```

## 7. creating, reading and extracting zip archives

```python
import zipfile
```

### creating

```python
file_list = ["file1.txt", "folder1/", "folder1/file1.txt"]

with zipfile.ZipFile('new.zip', 'w') as new_zip:
    for name in file_list:
        new_zip.write(name)
```

```python
os.remove("file1.txt")
os.remove("file3.txt")
os.remove("folder1/")
os.remove("backup/folder1_backup/")
os.remove("backup/")
os.remove("folder3")
```

### reading

```python
with zipfile.ZipFile('new.zip', 'r') as zipobj:
    print(zipobj.namelist(), "\n")

    bar_info = zipobj.getinfo('folder1/file1.txt')
    print(bar_info.filename)
    print(bar_info.date_time)
    print(bar_info.file_size)
    print(bar_info.compress_size)
```

### extracting

```python
data_zip = zipfile.ZipFile('new.zip', 'r')
```

```python
data_zip.extract('file1.txt')
os.listdir()
```
