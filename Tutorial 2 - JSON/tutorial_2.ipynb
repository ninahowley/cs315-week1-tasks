## 1. Writing JSON with Python

```python
import json
```

```python
food_ratings = {"organic dog food": 2, "human food": 10}
json.dumps(food_ratings)
```

```python
str(food_ratings)
```

```python
numbers_present = {1:True, 2: True, 3: False}
json.dumps(numbers_present)
```

```python
dog_id = 1
dog_name = "Frieda"
dog_registry = {dog_id: {"name": dog_name}}
json.dumps(dog_registry)
```

```python
json.dumps(True)
```

```python
json.dumps(["eating", "sleeping", "barking"])
```

```python
available_nums = {(1, 2): True, 3: False}
json.dumps(available_nums, skipkeys = True)
```

```python
toy_conditions = {"chew bone": 7, "ball": 3, "sock": -1}
json.dumps(toy_conditions, sort_keys=True)
```

### writing a json file with python

```python
cat_data = {
    "name": "Gizmo",
    "is_cat": True,
    "hobbies": ["sleeping", "playing"],
    "age": 1,
    "address": {
        "work": None,
        "home": ("Boston", "Massachusetts"),
    },
    "friends": [
        {
            "name": "Chupi",
            "hobbies": ["eating", "hunting"]
        },
        {
            "name": "Gatalita",
            "hobbies": ["eating", "sleeping"]
        },
    ],
}

with open("hello_gizmo.json", mode="w", encoding="utf-8") as write_file:
    json.dump(cat_data, write_file)
```

### reading a json file with python

```python
cat_registry = {1: {"name": "Gizmo"}}
cat_json = json.dumps(cat_registry)
cat_json
```

```python
new_cat_registry = json.loads(cat_json)
```

```python
new_cat_registry == cat_registry
```

```python
new_cat_registry
```

```python
cat_registry
```

### deserialize json data types

```python
cat_data_json = json.dumps(cat_data)
cat_data
```

```python
new_cat_data = json.loads(cat_data_json)
new_cat_data
```

```python
print(type(cat_data["address"]["home"]))
print(type(new_cat_data["address"]["home"]))
```

### open an external json file with python

```python
with open("hello_gizmo.json", mode="r", encoding="utf-8") as read_file:
    giz_data = json.load(read_file)

print(type(giz_data))
print(giz_data["name"])
```

```python
from hello_gizmo import cat_data

print(giz_data == cat_data)
print(type(giz_data["address"]["home"]))
print(type(cat_data["address"]["home"]))
```

## 2. interacting with json

### prettify json with python

```python
cat_friend = {
    "name": "Chupi",
    "age": 8
}
```

```python
print(json.dumps(cat_friend))
```

```python
print(json.dumps(cat_friend, indent=0))
```

```python
print(json.dumps(cat_friend, indent=2))
```

```python
with open("hello_gizmo.json", mode="w", encoding="utf-8") as write_file:
    json.dump(cat_data, write_file, indent = 2)
```

```python
with open("hello_gizmo.json", mode="r", encoding="utf-8") as input_file:
    original_json = input_file.read()

json_data = json.loads(original_json)
mini_json = json.dumps(json_data, indent=None, separators = (",", ":"))
with open("mini_gizmo.json", mode="w", encoding="utf-8") as output_file:
    output_file.write(mini_json)
```
